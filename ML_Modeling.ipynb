```python
#STEP-01:Import the Libraries

import numpy as np                                                                   # Import NumPy for numerical calculations
import pandas as pd                                                                  # Import Pandas for data handling
import matplotlib.pyplot as plt                                                      # Import Matplotlib for plotting graphs
import seaborn as sns                                                                # Import Seaborn for better visualizations
#-------------------------------------------------------------------------------------------------------------------
from ucimlrepo import fetch_ucirepo                                                  # Import dataset from UCI ML Repository                                                         
from sklearn.linear_model import LinearRegression                                    # Import Linear Regression model
from sklearn.impute import SimpleImputer                                             # Import tool to fill missing values
from sklearn.pipeline import Pipeline                                                # Import Pipeline to combine preprocessing and model steps
from sklearn.compose import ColumnTransformer                                        # Import transformer for preprocessing column
from sklearn.model_selection import train_test_split, GridSearchCV                   # Import data splitting and hyperparameter tuning
from sklearn.preprocessing import OneHotEncoder, PolynomialFeatures                  # Import encoding and polynomial feature tools
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score        # Import model evaluation metrics
sns.set_style(style = 'whitegrid')                                                   #To Set the background colour of Graph
```

```python
#STEP-02: Loading And Cleaning the Data (autompg dataset)
data = fetch_ucirepo(id = 9)
data
x = data.data.features.copy()  #it returs all nummerical features
y = data.data.targets['mpg']
x
```

```python
x.describe()
x.info()
```

```python
x['origin'].unique();
x['horsepower'].unique()
```

```python
#first we will transform the horse power and origin coloumns
x['horsepower'] = pd.to_numeric(x['horsepower'],errors = 'coerce')
x['origin'] = x['origin'].astype(str)
```

```python
x.info();
numerical_features = ['cylinders','displacement','horsepower','weight',
                      'accleration','model_year']
categorical_features = ['origin']
x = x[numerical_features+categorical_features]
x.info()
#x.isnull().sum()
```

```python
x.shape
```

```python
#Step-03:Quick Exploratory Data Analysis
df = x.copy()
df['mpg'] = y
#here we are mapping
```

```python
#we are gonna use sunplots() concept to divide plot into multiple plots
fig,axes = plt.subplots(1,3,figsize = (16,4))
#plot graphs as per axes
sns.histplot(df['mpg'],kde = False,ax = axes[0],color = 'black')
axes[0].set_title("MPG Disribution")
sns.scatterplot(x = df['weight'],y = df['mpg'],ax = axes[1],color = 'red')
axes[1].set_title("Weight VS MPG Distribution")
sns.scatterplot(x = df['horsepower'],y = df['mpg'],ax = axes[2],color = 'blue')
axes[2].set_title("HorsePower VS MPG Distribution")
```

```python
plt.figure(figsize=(16,4))
sns.heatmap(df.select_dtypes(include=np.number).corr(),
            annot=True,
            cmap='coolwarm',
            center=0)
plt.title("CORRELATION MATRIX")
plt.tight_layout()
plt.show()
```

```python
#print(x) , print(y)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print("\nTrain Test Split")
print("TRAINING DATA : ",len(x_train))
print("TESTING DATA : ",len(x_test))
print(x_train)
```

```python
def evaluate(model_name,actual,predicted):
    """This function returns MAE,MSE,RMSE,R2-Score"""
    mse=mean_squared_error(actual,predicted)
    return {'Mode Name': model_name,
            'MAE' : mean_absolute_error,
            'MSE' : mse,
            'RSME': np.sqrt(mse),
            'R2_SCORE' : r2_score(actual,predicted)}
results=[]
```

```python
base_line_predictions = np.full(len(y_test),y_train.mean())
#base_line_predictions

#in above case we are keeping a baseline predictions
results.append(evaluate("Mean Baseline",y_test,base_line_predictions))
results
```

```python
simple_model = LinearRegression()
#we fit the data to the model
simple_model.fit(x_train[['cylinders']],y_train)
simple_model_predictions = simple_model.predict(x_test[['cylinders']])
print(simple_model_predictions)
results.append(evaluate("Simple Linear Regression:Cylinders",y_test,
                        simple_model_predictions))
print(len(results))
print(results)
```
